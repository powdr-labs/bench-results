```python
import pandas as pd
```

```python
def prepare_data(data, time_column = "total_proof_time_ms"):
    data = data.copy()

    data.loc[:, "proof_time_s"] = (data[time_column] / 1000).round(2)

    mask = data["filename"].str.contains("apc000", na=False)
    row = data.loc[mask].drop_duplicates()
    software_proof_time = row["proof_time_s"].squeeze()
    software_proof_time

    data.loc[:, "faster_than_baseline"] = (software_proof_time / data["proof_time_s"]).round(2)
    return data[["filename", "num_segments", "proof_time_s", "faster_than_baseline"]]
```

# Manual + APC (with recursion)

```python
data = pd.read_csv("latest_nightly/reth/basic_metrics.csv")
```

```python
prepare_data(data, "total_proof_time_ms")
```

# Manual + APC (no recursion)

```python
data2 = pd.read_csv("basic_metrics.csv")
```

```python
prepare_data(pd.concat([data, data2]), "app_proof_time_ms")
```
